```python
import os
import warnings
import yaml
from pathlib import Path

PROJECT_DIR = Path().resolve()
PROJECT_ROOT_NAME = 'spectroscopy-regression'

while PROJECT_DIR.name != PROJECT_ROOT_NAME and PROJECT_DIR != PROJECT_DIR.parent:
    PROJECT_DIR = PROJECT_DIR.parent

if Path().resolve() != PROJECT_DIR:
    os.chdir(PROJECT_DIR)
```

```python
import pandas as pd
import os

input_dir = "data/NGC_combi.csv"      
columns_to_keep = ['O/Fe', 'F275W_abs', 'F336W_abs', 'F435W_abs', 'F606W_abs', 'F814W_abs', 
                   'F275W', 'F336W', 'F435W', 'F606W', 'F814W', 'age_Kruijssen', 'Fe/H']

df = pd.read_csv(input_dir, usecols=columns_to_keep)
for col in columns_to_keep:
    df[col] = pd.to_numeric(df[col], errors='coerce')

mask = (
    (df < -100).any(axis=1) |         
    df.isna().any(axis=1)             
)
df = df[~mask] 
df.to_csv('data/Photometry+Age+Metallicity+OFe.csv', index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load CSV file (replace 'your_file.csv' with your actual filename)
df = pd.read_csv('data/Na_Fe_training_data.csv')

# Specify the column name you want to inspect (replace 'column_name' with actual name)
column_name = 'Na/Fe'

# Create 3-panel figure
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Histogram
axes[0].hist(df[column_name], bins=30, edgecolor='black', alpha=0.7)
axes[0].set_title(f'Distribution of {column_name}')
axes[0].set_xlabel(column_name)
axes[0].set_ylabel('Frequency')

# Density plot
df[column_name].plot(kind='density', linewidth=2, ax=axes[1])
axes[1].set_title(f'Density Plot of {column_name}')
axes[1].set_xlabel(column_name)

# Scatter plot (index vs values)
axes[2].scatter(range(len(df)), df[column_name], alpha=0.6)
axes[2].set_title(f'Scatter Plot of {column_name}')
axes[2].set_xlabel('Index')
axes[2].set_ylabel(column_name)

plt.tight_layout()
plt.savefig('column_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```
